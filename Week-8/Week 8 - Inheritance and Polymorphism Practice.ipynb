# Inheritance in Python

Inheritance is the capability of one class to derive or inherit the properties from another class. The benefits of inheritance are: 
 
<ol><li>
It represents real-world relationships well.</li><li>
It provides reusability of a code, so that you don’t have to write the same code again and again. </li><li>It allows for adding more features to a class without modifying it.</li><li>
It is transitive in nature, which means that if class B inherits from another class A, then all the subclasses of B would automatically inherit from class A.</li><ol>

### A Python program to demonstrate inheritance

```python
class Person(object):

    # Constructor
    def __init__(self, name):
        self.name = name

    # To get name method
    def getname(self):
        return self.name

    # To check if this person is an employee
    def isEmployee(self):
        return False
    

# Inherited or Subclass (Note Person in bracket)
class Employee(Person):

    # Here employee method returns true
    def isEmployee(self):
        return True

# Object Instantiation
emp = Person("Segun Da-Silver") # An Object of Person
print(emp.getname(), emp.isEmployee())

emp = Employee("Omotayo Ayeni") # An Object of Employee
print(emp.getname(), emp.isEmployee())
```

### Subclassing (Calling constructor of parent class) 
<ul><li>A child class needs to identify which class is its parent class.</li><li> This can be done by mentioning the parent class name in the definition of the child class. 
Eg: class subclass_name (superclass_name): </li></ul>

```python
# Python code to demonstrate how parent constructors
# are called.

# parent class
class Person(object):

        # __init__ is known as the constructor
        def __init__(self, name, idnumber):
            self.name = name
            self.idnumber = idnumber
            
        def display(self):
            print("Name:", self.name)
            print("ID No.:", self.idnumber)
            print("Salary:", self.salary)

# child class
class Employee(Person):
    def __init__(self, name, idnumber, salary, post):
        self.salary = salary
        self.post = post

        # invoking the __init__ of the parent class
        Person.__init__(self, name, idnumber)

                
# creation of an object variable or an instance
a = Employee('Adedoyin Adeniji', 886012, 200000, "Intern")

# calling a function of the class Person using its instance
a.display()
```

### Note that:
<ul><li>In Python, every class inherits from a built-in basic class called ‘object’. The constructor i.e. the ‘__init__’ function of a class is invoked when we create an object variable or an instance of the class.</li><li>
The variables defined within __init__() are called as the instance variables or objects. Hence, ‘name’ and ‘idnumber’ are the objects of the class Person. Similarly, ‘salary’ and ‘post’ are the objects of the class Employee.</li><li> Since the class Employee inherits from class Person, ‘name’ and ‘idnumber’ are also the objects of class Employee.</li><li>
If you forget to invoke the __init__() of the parent class then its instance variables would not be available to the child class. </li></ul>

```python
# Python program to demonstrate error if we
# forget to invoke __init__() of the parent.

class A:
    # class attribute
    surname= "Moru"
    
    def __init__(self, n = 'Emmanuel Abbah'):
        #instance attribute
        self.name = n
        
class B(A):
    def __init__(self, roll):
        self.roll = roll
        
    
# Object Instance
project = B(23)
print (project.A)
```

## Types of Inheritance in Python
There are four types of inheritance in Python. The type of Inheritance depends upon the number of child and parent classes involved. : 

#### Single Inheritance:
<ul><li> Single inheritance enables a derived class to inherit properties from a single parent class, thus enabling code reusability and the addition of new features to existing code.</li></ul>

```python
# Python program to demonstrate
# single inheritance


# Base class
class parents: 
    def func1(self):
        print("This function is in parent class.")

# Derived class
class Child(parents):
    def func2(self):
        print("This function is in child class.")

# Object Instance
object = Child()
object.func1()
object.func2()
```

#### Multiple Inheritance:
<ul><li>When a class can be derived from more than one base class this type of inheritance is called multiple inheritance. </li><li>In multiple inheritance, all the features of the base classes are inherited into the derived class. </li></ul>
 

```python
# Python program to demonstrate
# multiple inheritance


# Base class1
class Mother: 
    mothername = "Caroline Aina"
    def mother(self):
        print(self.mothername)

# Base class2
class Father:
    fathername = "Paul Abiodun"
    def father(self):
        print(sef.fathername)

# Derived class
class Son(Mother, Father):
    def parents(self):
        print("Father :", self.fathername)
        print("Mother :", self.mothername)

# Object Instance
s1 = Son()
s1.parents()
```

#### Multilevel Inheritance
<ul><li>In multilevel inheritance, features of the base class and the derived class are further inherited into the new derived class.</li><li> This is similar to a relationship representing a child and grandfather.</li></ul> 

```python
# Python program to demonstrate
# multilevel inheritance

# Base class
class Grandfather:

    def __init__(self, grandfathername):
        self.grandfathername = grandfathername

# Intermediate class
class Father(Grandfather):
    def __init__(self, fathername, grandfathername):
        self.fathername = fathername
        self.grandfathername = grandfathername

        # invoking constructor of Grandfather class
        Grandfather.__init__(self, grandfathername)

# Derived class
class Son(Father):
    def __init__(self,sonname, fathername, grandfathername):
        self.sonname = sonname
        

        # invoking constructor of Father class
        Father.__init__(self, fathername, grandfathername)

    def print_name(self):
        print('Grandfather name :', self.grandfathername)
        print("Father name :", self.fathername)
        print("Son name :", self.sonname)

# Object Instance
s1 = Son('Gbenga', 'Oluwafemi', 'Adeoke')
print(s1.grandfathername)
s1.print_name()
```

#### Hierarchical Inheritance:
<ul><li>When more than one derived classes are created from a single base this type of inheritance is called hierarchical inheritance.</li><li> In this program, we have a parent (base) class and two child (derived) classes.</li></ul>

```python
# Python program to demonstrate
# Hierarchical inheritance


# Base class
class Parent:
    def func1(self):
        print("This function is in parent class.")

# Derived class1
class Child1(Parent):
    def func2(self):
        print("This function is in child 1.")

# Derivied class2
class Child2(Parent):
    def func3(self):
        print("This function is in child 2.")

# Object Instantiation
object1 = Child1()
object2 = Child2()
object1.func1()
object1.func2()
object2.func1()
object2.func3()
```

#### Hybrid Inheritance:
<ul><li>Inheritance consisting of multiple types of inheritance is called hybrid inheritance.</li></ul>

```python
# Python program to demonstrate
# hybrid inheritance


class School():
    def func1(self):
        print("This function is in school.")

class Student1(School):
    def func2(self):
        print("This function is in student 1. ")

class Student2(School):
    def func3(self):
        print("This function is in student 2.")

class Student3(Student1, School):
    def func4(self):
        print("This function is in student 3.")

# Object Instantiation
object = Student3()
object.func1()
object.func2()

```

# Polymorphism in Python

<ul><li>In programming, Polymorphism is a concept of OOP.</li><li>The word polymorphism means having many forms.</li><li>
    It enables using a single interface with the input of different data types, different classes or maybe for a different number of inputs.</li><li>Polymorphism means the same function name (but different signatures) being used for different types.</li></ul>

### Inbuilt polymorphic functions

```python
# Python program to demonstrate in-built polymorphic functions

# len() being used for a string
print(len("Department of Computer Science"))

# len() being used for a list
print(len([10, 20, 30, 40, 50, 60, 70]))
```

### User-defined polymorphic functions : 

```python
# A simple Python function to demonstrate
# Polymorphism

def add(x, y, z, w = 6): 
    return x + y 

# Driver code
print(add(2, 3, 4))
print(add(2, 3, 4))
print(add(2, 3, 5, 6))
```

### Polymorphism with class methods: 
<ul><li>The code below shows how Python can use two different class types, in the same way.</li><li> We create a for loop that iterates through a tuple of objects. Then call the methods without being concerned about which class type each object is. We assume that these methods actually exist in each class.</li></ul> 

```python
class Nigeria():
    def capital(self):
        print("Lagos is the capital of Nigeria.")

    def language(self):
        print("English is the official language of Nigeria.")

    def type(self):
        print("Nigeria is the giant of Africa.")

class Togo():
    def capital(self):
        print("Lome is the capital of Togo.")

    def language(self):
        print("French is the primary language of Togo.")

    def type(self):
        print("Togo is known for its palm-lined beaches and hilltop villages.")

obj_naija = Nigeria()
obj_togo = Togo()
for country in (obj_naija, obj_togo):
    country.capital()
    country.language()
    country.type()
```

### Polymorphism with Inheritance:
<ul><li>In Python, Polymorphism lets us define methods in the child class that have the same name as the methods in the parent class. In inheritance, the child class inherits the methods from the parent class.</li><li> However, it is possible to modify a method in a child class that it has inherited from the parent class.</li><li> This is particularly useful in cases where the method inherited from the parent class doesn’t quite fit the child class. In such cases, we re-implement the method in the child class.</li><li> This process of re-implementing a method in the child class is known as <b>Method Overriding. </b> </li></ul>

```python
class Bird:
    def intro(self):
        print("There are many types of birds.")

    def flight(self):
        print("Most of the birds can fly but some cannot.")

class sparrow(Bird):
    def flight(self):
        print("Sparrows can fly.")

class ostrich(Bird):
    def flight(selfie):
        print("Ostriches cannot fly.")

obj_bird = Bird()
obj_spr = sparrow()
obj_ost = ostrich()

obj_bird.intro()
obj_bird.flight()

obj_spr.intro()
obj_spr.flight()

obj_ost.intro()
obj_ost.flight()
```

    Ploymorphism can be used to overide exclusive services.</li><ul>

```python
import tkinter as tk
from tkinter import font as tkfont


class Zenith:

    BANK_NAME = "Zenith Bank Nigeria"

    _MUTUAL_SERVICES = [
        "Lines of Credit",
        "Investment Management and Accounts",
        "Insurance",
    ]

    def mutual_services(self) -> list[str]:
        return self._MUTUAL_SERVICES

    def unique_services(self) -> list[str]:
        return []

    def all_services(self) -> list[str]:
        return self.unique_services() + [
            s for s in self.mutual_services()
            if s not in self.unique_services()
        ]

    def division_name(self) -> str:
        return "Zenith Bank"

    def division_color(self) -> str:
        return "#e63946"


class RetailBanking(Zenith):

    def division_name(self) -> str:
        return "Retail Banking Division"

    def division_color(self) -> str:
        return "#e63946"

    def unique_services(self) -> list[str]:
        return [
            "Retirement and Education Accounts",
            "Loans and Mortgages",
            "Checking and Saving",
        ]

    def all_services(self) -> list[str]:
        return self.mutual_services() + self.unique_services()


class GlobalBanking(Zenith):

    def division_name(self) -> str:
        return "Global Banking Division"

    def division_color(self) -> str:
        return "#1d3557"

    def mutual_services(self) -> list[str]:
        return []

    def unique_services(self) -> list[str]:
        return [
            "Multi-Currency Management Services and Products",
            "Foreign Currency Accounts",
            "Foreign Currency Credit Cards",
            "Transborder Advisory Services",
            "Liquidity Management",
        ]

    def all_services(self) -> list[str]:
        return self.unique_services()

class CommercialBanking(Zenith):

    def division_name(self) -> str:
        return "Commercial Banking Division"

    def division_color(self) -> str:
        return "#2a9d8f"      

    def unique_services(self) -> list[str]:
        return ["Advisory Services"]

    def all_services(self) -> list[str]:
        return self.mutual_services() + self.unique_services()


EMPLOYEES: dict[str, str] = {
    "mary":   "Retail Banking",
    "agatha": "Global Banking",
    "noel":   "Commercial Banking",
}

DIVISIONS: dict[str, Zenith] = {
    "Retail Banking":    RetailBanking(),
    "Global Banking":    GlobalBanking(),
    "Commercial Banking": CommercialBanking(),
}


class App(tk.Tk):
    BG      = "#f8f5f0"
    WHITE   = "#ffffff"
    RED     = "#e63946"
    NAVY    = "#1d3557"
    TEAL    = "#2a9d8f"
    DARK    = "#1a1a2e"
    MUTED   = "#6c757d"
    BORDER  = "#dee2e6"
    GOLD    = "#f4a261"

    FONT_BRAND  = ("Georgia", 22, "bold")
    FONT_TAGLINE= ("Georgia", 10, "italic")
    FONT_LABEL  = ("Helvetica Neue", 9, "bold")
    FONT_INPUT  = ("Helvetica Neue", 13)
    FONT_BTN    = ("Helvetica Neue", 11, "bold")
    FONT_DIV    = ("Georgia", 15, "bold")
    FONT_SERVICE= ("Helvetica Neue", 11)
    FONT_TINY   = ("Helvetica Neue", 9)

    def __init__(self):
        super().__init__()
        self.title("Zenith Bank — Division Services")
        self.configure(bg=self.BG)
        self.resizable(False, False)
        self._build_ui()
        self._center(520, 680)


    def _build_ui(self):
        hdr = tk.Frame(self, bg=self.RED, pady=0)
        hdr.pack(fill="x")

        stripe = tk.Frame(hdr, bg=self.GOLD, height=5)
        stripe.pack(fill="x")

        hdr_inner = tk.Frame(hdr, bg=self.RED, pady=20)
        hdr_inner.pack(fill="x")

        tk.Label(hdr_inner, text="🏦  Zenith Bank Nigeria",
                 bg=self.RED, fg=self.WHITE,
                 font=self.FONT_BRAND).pack()
        tk.Label(hdr_inner,
                 text="Your Aptitude for Excellence",
                 bg=self.RED, fg="#f9c8cc",
                 font=self.FONT_TAGLINE).pack(pady=(2, 0))

        tk.Frame(hdr, bg=self.GOLD, height=5).pack(fill="x")

        card = tk.Frame(self, bg=self.WHITE,
                        highlightthickness=1,
                        highlightbackground=self.BORDER)
        card.pack(fill="x", padx=28, pady=20)

        ci = tk.Frame(card, bg=self.WHITE, padx=26, pady=22)
        ci.pack(fill="x")

        tk.Label(ci, text="EMPLOYEE NAME",
                 bg=self.WHITE, fg=self.MUTED,
                 font=self.FONT_LABEL, anchor="w").pack(fill="x")

        ef = tk.Frame(ci, bg=self.WHITE,
                      highlightthickness=1,
                      highlightbackground=self.BORDER)
        ef.pack(fill="x", pady=(4, 14))

        self.name_var = tk.StringVar()
        tk.Entry(ef, textvariable=self.name_var,
                 font=self.FONT_INPUT,
                 bg=self.WHITE, fg=self.DARK,
                 insertbackground=self.RED,
                 relief="flat", bd=0).pack(
                     fill="x", ipady=9, padx=10)

        tk.Label(ci, text="DIVISION",
                 bg=self.WHITE, fg=self.MUTED,
                 font=self.FONT_LABEL, anchor="w").pack(fill="x")

        self.div_var = tk.StringVar(value="Retail Banking")
        btn_row = tk.Frame(ci, bg=self.WHITE)
        btn_row.pack(fill="x", pady=(4, 16))

        self._div_btns: dict[str, tk.Label] = {}
        for div in DIVISIONS:
            short = div.replace(" Banking", "").replace(" ", "\n")
            lbl = tk.Label(btn_row, text=short,
                           bg=self.BORDER, fg=self.DARK,
                           font=("Helvetica Neue", 10, "bold"),
                           padx=12, pady=8, cursor="hand2",
                           relief="flat", width=10)
            lbl.pack(side="left", padx=(0, 8))
            lbl.bind("<Button-1>", lambda e, d=div: self._select_div(d))
            self._div_btns[div] = lbl

        self._select_div("Retail Banking")

        tk.Button(ci, text="VIEW DIVISION SERVICES",
                  command=self._lookup,
                  font=self.FONT_BTN,
                  bg=self.RED, fg=self.WHITE,
                  activebackground="#c1121f",
                  activeforeground=self.WHITE,
                  relief="flat", cursor="hand2",
                  pady=11).pack(fill="x")

        self.result_outer = tk.Frame(self, bg=self.BG)
        self.result_outer.pack(fill="both", expand=True,
                               padx=28, pady=(0, 20))

        self._draw_placeholder()
        tk.Label(self,
                 text="© 2025 Zenith Bank Nigeria Plc  ·  RC 84402",
                 bg=self.BG, fg=self.BORDER,
                 font=self.FONT_TINY).pack(pady=(0, 10))

    def _select_div(self, division: str):
        self.div_var.set(division)
        colors = {
            "Retail Banking":    self.RED,
            "Global Banking":    self.NAVY,
            "Commercial Banking": self.TEAL,
        }
        for d, lbl in self._div_btns.items():
            if d == division:
                lbl.configure(bg=colors[d], fg=self.WHITE)
            else:
                lbl.configure(bg=self.BORDER, fg=self.DARK)

    def _lookup(self):
        name = self.name_var.get().strip()
        division = self.div_var.get()

        if not name:
            self._show_error("Please enter an employee name.")
            return
        known_div = EMPLOYEES.get(name.lower())
        div_obj = DIVISIONS[division]
        if known_div and known_div != division:
            self._show_error(
                f"{name.title()} works in the {known_div} Division,\n"
                f"not in {division}."
            )
            return

        if not known_div and name.lower() not in EMPLOYEES:
            employee_note = f"Guest / Visitor — {division}"
        else:
            employee_note = f"{name.title()} · {div_obj.division_name()}"

        self._draw_services(employee_note, div_obj)


    def _clear_result(self):
        for w in self.result_outer.winfo_children():
            w.destroy()

    def _draw_placeholder(self):
        self._clear_result()
        ph = tk.Frame(self.result_outer, bg=self.WHITE,
                      highlightthickness=1,
                      highlightbackground=self.BORDER)
        ph.pack(fill="both", expand=True)
        tk.Label(ph,
                 text="Enter an employee name\nand select a division to view services.",
                 bg=self.WHITE, fg=self.MUTED,
                 font=("Georgia", 12, "italic"),
                 justify="center").pack(expand=True)

    def _draw_services(self, employee_note: str, div: Zenith):
        self._clear_result()
        color = div.division_color()
        banner = tk.Frame(self.result_outer, bg=color, pady=14)
        banner.pack(fill="x")
        tk.Label(banner, text=div.division_name(),
                 bg=color, fg=self.WHITE,
                 font=self.FONT_DIV).pack()
        tk.Label(banner, text=employee_note,
                 bg=color, fg="#d0d8e8",
                 font=("Helvetica Neue", 10)).pack(pady=(2, 0))
        svc_frame = tk.Frame(self.result_outer, bg=self.WHITE,
                             highlightthickness=1,
                             highlightbackground=self.BORDER)
        svc_frame.pack(fill="x")

        mutual = set(div.mutual_services())
        unique = set(div.unique_services())

        all_svcs = div.all_services()

        legend = tk.Frame(svc_frame, bg="#f8f8f8", pady=6)
        legend.pack(fill="x", padx=16, pady=(10, 4))

        for dot, label in [("●", "Mutual service"), ("◆", "Exclusive service")]:
            row = tk.Frame(legend, bg="#f8f8f8")
            row.pack(side="left", padx=(0, 20))
            tk.Label(row, text=dot, bg="#f8f8f8",
                     fg=self.GOLD if dot == "●" else color,
                     font=("Helvetica Neue", 10)).pack(side="left")
            tk.Label(row, text=f"  {label}", bg="#f8f8f8",
                     fg=self.MUTED,
                     font=self.FONT_TINY).pack(side="left")

        tk.Frame(svc_frame, bg=self.BORDER, height=1).pack(fill="x", padx=16)

        for i, svc in enumerate(all_svcs):
            row_bg = self.WHITE if i % 2 == 0 else "#fafafa"
            is_mutual = svc in mutual
            dot   = "●" if is_mutual else "◆"
            dot_c = self.GOLD if is_mutual else color
            tag   = "Mutual" if is_mutual else "Exclusive"
            tag_bg= "#fff8e7" if is_mutual else self._lighten(color)
            tag_fg= "#996600" if is_mutual else color

            row = tk.Frame(svc_frame, bg=row_bg, pady=10)
            row.pack(fill="x", padx=16)

            tk.Label(row, text=dot, bg=row_bg, fg=dot_c,
                     font=("Helvetica Neue", 14)).pack(side="left")
            tk.Label(row, text=f"  {svc}", bg=row_bg, fg=self.DARK,
                     font=self.FONT_SERVICE).pack(side="left")
            tk.Label(row, text=tag,
                     bg=tag_bg, fg=tag_fg,
                     font=self.FONT_TINY,
                     padx=6, pady=2,
                     relief="flat").pack(side="right")

        foot = tk.Frame(svc_frame, bg="#f0f0f0", pady=8)
        foot.pack(fill="x")
        tk.Label(foot,
                 text=f"{len(all_svcs)} services total  ·  "
                      f"{len([s for s in all_svcs if s in mutual])} mutual  ·  "
                      f"{len([s for s in all_svcs if s in unique])} exclusive",
                 bg="#f0f0f0", fg=self.MUTED,
                 font=self.FONT_TINY).pack()

    def _show_error(self, msg: str):
        self._clear_result()
        err = tk.Frame(self.result_outer, bg="#fff0f0",
                       highlightthickness=1,
                       highlightbackground="#f5c6cb")
        err.pack(fill="both", expand=True)
        tk.Label(err, text="⚠  " + msg,
                 bg="#fff0f0", fg=self.RED,
                 font=("Helvetica Neue", 12),
                 wraplength=420, justify="center").pack(expand=True)

    @staticmethod
    def _lighten(hex_color: str) -> str:
        r = int(hex_color[1:3], 16)
        g = int(hex_color[3:5], 16)
        b = int(hex_color[5:7], 16)
        r = r + (255 - r) * 4 // 5
        g = g + (255 - g) * 4 // 5
        b = b + (255 - b) * 4 // 5
        return f"#{r:02x}{g:02x}{b:02x}"


    def _center(self, w: int, h: int):
        self.update_idletasks()
        x = (self.winfo_screenwidth()  - w) // 2
        y = (self.winfo_screenheight() - h) // 2
        self.geometry(f"{w}x{h}+{x}+{y}")


if __name__ == "__main__":
    App().mainloop()
```
