```python
input_1 = int(input("How many years of experience do you have? "))
input_2 = int(input("How old are you? "))

if input_1 > 25 and input_2 >= 55:
    print("Your Annual Tax Revenue (ATR) is N5,600,000")
elif input_1 > 20 and input_2 >= 45:
    print("Your Annual Tax Revenue (ATR) is N4,480,000")
elif input_1 > 10 and input_2 >= 35:
    print("Your Annual Tax Revenue (ATR) is N1,500,000")
elif input_1 < 10 and input_2 < 35:
    print("Your Annual Tax Revenue (ATR) is N550,000")
else:
    print("Invalid input")



    
    
```
