# Encapsulation in Python

<ul><li>Encapsulation is the concept of bundling data and methods within a single unit. For example, when you create a class, it means you are implementing encapsulation. A class is an example of encapsulation as it binds all the data members (instance variables) and methods into a single unit.</li><li>

Using encapsulation, we can hide an object’s internal representation from the outside. This is called information hiding.</li><li>

Encapsulation allows us to restrict accessing variables and methods directly and prevent accidental data modification by creating private data members and methods within a class.</li><li>

Encapsulation is a way to restrict access to methods and variables from outside of class. Whenever working with the class and dealing with sensitive data, providing access to all variables used within the class is not a good choice.</li></ul>

```python
# A program that displays employees' information

class Employee:
    
    # constructor
    def __init__(self, name, salary, project):        
        # instance variables
        self.name = name
        self.salary = salary
        self.project = project

    # method to display employee's details
    def show(self):        
        # accessing public instance variables
        print("\nName: " , self.name, '\nSalary:', "N"+str(self.salary))

    # method
    def work(self):
        print(self.name, 'is working on', self.project)


name = input("Enter your name: ")
salary = int(input("How much do you earn: "))
project = input("What project are you working on: ")

# creating object of a class
emp = Employee(name, salary, project)

# calling public method of the class
emp.show()
emp.work()
```

## Access Modifiers in Python

<ul><li>Encapsulation can be achieved by declaring the data members and methods of a class either as private or protected.</li><li> In Python, there is no direct access modifiers like public, private, and protected. </li><li>This can achieve by using single underscore and double underscores.</li></ul>

Access modifiers limit access to the variables and methods of a class. Python provides three types of access modifiers private, public, and protected.

<ul><li><b>Public Instance Variable:</b> Accessible anywhere from outside oclass.</li><li>
    <b>Private Instance Variable:</b> Accessible within the class.</li><li>
    <b>Protected Instance Variable:</b> Accessible within the class and its sub-classes.</li></ul>

### Public Instance Variable
Public instance variables are accessible within and outside of a class. All member variables of the class are by default public.

```python
class Employee:
    
    # constructor
    def __init__(self, name, salary):        
        # public instance variables
        self.name = name
        self.salary = salary

    # public instance methods
    def show(self):
        # accessing public instance variables
        print("Name: ", self.name, 'Salary:', "N%s"%(self.salary))

# creating object of a class
emp = Employee('Abdurrahman', 500000)

# accessing public instance variables
print("Name: ", emp.name, 'Salary:', "N{}".format(emp.salary))

# calling public method of the class
emp.show()
```

### Private Instance Variable
<ul><li>Protect variables in the class by marking them private. To define a private variable add two underscores as a prefix at the start of a variable name.</li><li>

Private instance variables are accessible only within the class, and we can’t access them directly from the class objects.</li></ul>

```python
class Employee:
    
    # constructor
    def __init__(self, name, salary):        
        # public instance variable
        self.name = name
        
        # private variable
        self.salary = salary

# creating object of a class
emp = Employee('Ugochi Mbaekwe', 10000)

# accessing public instance variables
print("Name: ", emp.name)

# accessing private instance variable
print('Salary:', emp.salary)
```

To access private members from outside of a class using the following two approaches

<ul><li>Create public method to access private members</li><li>
Use name mangling</li></ul>

#### Access Private member outside of a class using an instance method

```python
class Employee:
    
    # constructor
    def __init__(self, name, salary):        
        # public data member
        self.name = name
        
        # private member
        self.salary = salary

    # public instance methods
    def show(self):
        # private members are accessible from a class
        print("Name: ", self.name, '\nSalary:', self.salary)

# creating object of a class
emp = Employee('Chikaodi Chinaka', 250000)

# calling public method of the class
emp.show()
```

#### Name Mangling to access private members
<ul><li>Private and protected variables can be directly accessed from outside of a class through name mangling.</li><li> The name mangling is created on an identifier by adding two leading underscores and one trailing underscore, like this <b>_classname__dataMember</b>, where <b><i>classname</i></b> is the current class, and data member is the private variable name.</li></ul>

```python
class Employee:
    # constructor
    def __init__(self, name, salary):
        # public instance variable
        self.name = name
        
        # private variable
        self.salary = salary

# creating object of a class
emp = Employee('Oyindamola Apampa', 900000)

#direct access to public instance variable
print('Name:', emp.name)

# direct access to private instance variable using name mangling
print('Salary:', "N"+str(emp.salary))
```

### Protected Instance Variable
<ul><li>Protected instance variables are accessible within the class and also available to its sub-classes. </li><li>To define a protected variable, prefix the variable name with a single underscore <b>_</b>.</li><li>

Protected instance variables are used when you implement inheritance and want to allow data members access to only child classes.</li></ul>

```python
# base class
class Company:
    # base constructor
    def __init__(self):
        # Protected instance variable
        self.project = "Blockchain Development"

# child class
class Employee(Company):
    # child constructor
    def __init__(self, name):
        self.name = name
        
        # invoke base constructor
        Company.__init__(self)

    def show(self):
        print("Employee name :", self.name)
        
        # Accessing protected instance variable in child class
        print("Working on project :", self.project)

c = Employee("Prince Ibekwe")
c.show()

# Direct access to protected instance variable
print('Project:', c.project)
```

## Getters and Setters in Python
<ul><li>To implement proper encapsulation in Python, setters and getters can be used.</li><li> The primary purpose of using getters and setters in object-oriented programs is to ensure data encapsulation.</li><li> Use the getter method to access instance variables and the setter methods to modify the instance variables.</li></ul>

In Python, private variables are not hidden fields like in other programming languages. The getters and setters methods are often used when:

<ul><li>When we want to avoid direct access to private variables</li><li>
To add validation logic for setting a value</li></ul>

```python
class Student:
    def __init__(self, name, age):
        # public instance variable
        self.name = name
        # private instance variable
        self.age = age

    # getter method
    def get_age(self):
        return self.age

    # setter method
    def set_age(self, age):
        self.age = age

stud = Student('Michael Illoba', 34)

# retrieving age using getter
print('Name:', stud.name, "\nAge:", stud.get_age())

# changing age using setter
stud.set_age(26)

# retrieving age using getter
print('\n\nName:', stud.name, "\nAge:", stud.get_age())
```

```python
class Student:
    # contructor
    def  __init__(self, name, roll_no, age):
        # public instance variable
        self.name = name
        # private instance variables to restrict access
        # avoid direct data modification
        self.roll_no = roll_no
        self.age = age

    def show(self):
        print('Student Details:', self.name, self.roll_no)

    # getter methods
    def get_roll_no(self):
        return self.roll_no

    # setter method to modify instance varaible
    # condition to allow data modification with rules
    def set_roll_no(self, number):
        if number > 50:
            print('Invalid roll no. Please set correct roll number')
        else:
            self.roll_no = number

# object instanc=tiation
info = Student('David Usim', 10, 15)

# before Modify
info.show()

# changing roll number using setter
info.set_roll_no(120)


info.set_roll_no(25)
info.show()
```

You have been contracted by the Registrar of Pan-Atlantic University (PAU) as an expert OOP developer to access the Student Information System (SIS) of PAU Student Council, inorder to classify the students grades, according to their age, into 3 categories; the pirates, the yankees and the bulls.

Should you choose to accept this task, develop an OOP program that reads data from the SIS.csv file as attached, following the instructions below: 

<b>Instructions:</b>
    <ul><li>If the student age is greater than 14 and less than 18, create a .csv file for that category called <b>The_Pirates</b> and display it.</li><li>
    If the student age is greater than 18 and less than 22, create a file for that category called <b>The_Yankees</b> and display it.</li><li>
    If the student age is greater than 22 and less than 25 create a file for that category called <b>The_Bulls</b> and display it.</li><li>
    If the student age is greater than 25 <b>encapsulate the request</b> and use a getter method to return an error message".</li></ul>


```python
import tkinter as tk
from tkinter import ttk, filedialog, messagebox
import pandas as pd
import os

class StudentRecord:

    def __init__(self, mat_no: int, name: str, age: int, grade: int):
        self._mat_no = mat_no
        self._name   = name
        self._age    = age
        self._grade  = grade

    def get_mat_no(self) -> int:  return self._mat_no
    def get_name(self)   -> str:  return self._name
    def get_age(self)    -> int:  return self._age
    def get_grade(self)  -> int:  return self._grade

    def to_dict(self) -> dict:
        return {
            "MatNo": self._mat_no,
            "Name":  self._name,
            "Age":   self._age,
            "Grade": self._grade,
        }


class SISClassifier:
    CATEGORIES = {
        "The_Pirates": lambda age: 14 < age < 18,
        "The_Yankees": lambda age: 18 < age < 22,
        "The_Bulls":   lambda age: 22 < age < 25,
    }

    BOUNDARY_AGES = {18, 22, 25}

    def __init__(self, csv_path: str, output_dir: str):
        self._csv_path   = csv_path
        self._output_dir = output_dir
        self._students: list[StudentRecord] = []
        self._groups: dict[str, list[StudentRecord]] = {k: [] for k in self.CATEGORIES}
        self._restricted: list[StudentRecord] = []
        self._boundary:   list[StudentRecord] = []
        self._error_message: str = ""

    def _load(self):
        df = pd.read_csv(self._csv_path)
        for _, row in df.iterrows():
            s = StudentRecord(
                int(row["MatNo"]), str(row["Name"]),
                int(row["Age"]),   int(row["Grade"])
            )
            self._students.append(s)

    def _classify(self):
        for s in self._students:
            age = s.get_age()
            matched = False
            for category, rule in self.CATEGORIES.items():
                if rule(age):
                    self._groups[category].append(s)
                    matched = True
                    break
            if not matched:
                if age in self.BOUNDARY_AGES:
                    self._boundary.append(s)
                elif age > 25:
                    self._restricted.append(s)

        if self._restricted:
            names = ", ".join(s.get_name() for s in self._restricted)
            self._error_message = (
                f"ACCESS RESTRICTED: {len(self._restricted)} student(s) "
                f"aged above 25 cannot be classified — {names}."
            )

    def _write_csvs(self):
        os.makedirs(self._output_dir, exist_ok=True)
        for category, members in self._groups.items():
            if members:
                path = os.path.join(self._output_dir, f"{category}.csv")
                pd.DataFrame([s.to_dict() for s in members]).to_csv(
                    path, index=False
                )

    def run(self):
        self._load()
        self._classify()
        self._write_csvs()

    def get_group(self, category: str) -> list[StudentRecord]:
        return self._groups.get(category, [])

    def get_restricted_error(self) -> str:
        return self._error_message

    def get_boundary(self) -> list[StudentRecord]:
        return self._boundary

    def get_all_students(self) -> list[StudentRecord]:
        return self._students

class App(tk.Tk):
    BG      = "#0f172a"
    PANEL   = "#1e293b"
    BORDER  = "#334155"
    TEXT    = "#e2e8f0"
    MUTED   = "#94a3b8"
    WHITE   = "#ffffff"

    CAT_COLORS = {
        "The_Pirates": {"bg": "#7c3aed", "lt": "#ede9fe", "icon": "☠"},
        "The_Yankees": {"bg": "#0369a1", "lt": "#e0f2fe", "icon": "⚾"},
        "The_Bulls":   {"bg": "#b91c1c", "lt": "#fee2e2", "icon": "🐂"},
    }

    FONT_BRAND  = ("Georgia", 20, "bold")
    FONT_SUB    = ("Georgia", 9, "italic")
    FONT_CAT    = ("Helvetica Neue", 13, "bold")
    FONT_BTN    = ("Helvetica Neue", 11, "bold")
    FONT_TABLE  = ("Courier New", 10)
    FONT_LABEL  = ("Helvetica Neue", 9, "bold")
    FONT_TINY   = ("Helvetica Neue", 9)
    FONT_ERR    = ("Helvetica Neue", 10, "bold")

    def __init__(self):
        super().__init__()
        self.title("PAU SIS — Student Grade Classifier")
        self.configure(bg=self.BG)
        self.resizable(True, True)
        self.classifier: SISClassifier | None = None
        self._build_ui()
        self._center(820, 700)

    def _build_ui(self):
        hdr = tk.Frame(self, bg="#1e1b4b", pady=18)
        hdr.pack(fill="x")
        tk.Label(hdr, text="🎓  Pan-Atlantic University",
                 bg="#1e1b4b", fg="#a5b4fc",
                 font=self.FONT_BRAND).pack()
        tk.Label(hdr, text="Student Information System  ·  Grade Classifier",
                 bg="#1e1b4b", fg="#6366f1",
                 font=self.FONT_SUB).pack(pady=(2, 0))
        ctrl = tk.Frame(self, bg=self.PANEL, pady=12, padx=20)
        ctrl.pack(fill="x")

        self.path_var = tk.StringVar(value="No file selected")
        tk.Label(ctrl, textvariable=self.path_var,
                 bg=self.PANEL, fg=self.MUTED,
                 font=self.FONT_TINY, anchor="w").pack(side="left", expand=True, fill="x")

        tk.Button(ctrl, text="📂  Load SIS.csv",
                  command=self._load_file,
                  font=self.FONT_BTN,
                  bg="#4f46e5", fg=self.WHITE,
                  activebackground="#4338ca",
                  activeforeground=self.WHITE,
                  relief="flat", cursor="hand2",
                  padx=14, pady=6).pack(side="right", padx=(8, 0))

        self.run_btn = tk.Button(ctrl, text="⚡  Classify Students",
                  command=self._run_classifier,
                  font=self.FONT_BTN,
                  bg="#059669", fg=self.WHITE,
                  activebackground="#047857",
                  activeforeground=self.WHITE,
                  relief="flat", cursor="hand2",
                  padx=14, pady=6, state="disabled")
        self.run_btn.pack(side="right")

        tk.Frame(self, bg=self.BORDER, height=1).pack(fill="x")

        style = ttk.Style(self)
        style.theme_use("clam")
        style.configure("Dark.TNotebook",
                        background=self.BG,
                        borderwidth=0)
        style.configure("Dark.TNotebook.Tab",
                        background=self.PANEL,
                        foreground=self.MUTED,
                        padding=[14, 8],
                        font=("Helvetica Neue", 10, "bold"))
        style.map("Dark.TNotebook.Tab",
                  background=[("selected", "#4f46e5")],
                  foreground=[("selected", self.WHITE)])

        self.nb = ttk.Notebook(self, style="Dark.TNotebook")
        self.nb.pack(fill="both", expand=True, padx=0, pady=0)

        self._tabs: dict[str, tk.Frame] = {}
        self._trees: dict[str, ttk.Treeview] = {}

        tab_defs = [
            ("All Students",  "All Students"),
            ("☠  Pirates",    "The_Pirates"),
            ("⚾  Yankees",   "The_Yankees"),
            ("🐂  Bulls",     "The_Bulls"),
            ("⚠  Restricted", "restricted"),
        ]

        for label, key in tab_defs:
            frame = tk.Frame(self.nb, bg=self.BG)
            self.nb.add(frame, text=f"  {label}  ")
            self._tabs[key] = frame
            self._build_tab(frame, key)

        
        self.status_var = tk.StringVar(value="Load a SIS.csv file to begin.")
        status = tk.Frame(self, bg="#0f172a", pady=5)
        status.pack(fill="x", side="bottom")
        tk.Label(status, textvariable=self.status_var,
                 bg="#0f172a", fg=self.MUTED,
                 font=self.FONT_TINY).pack(side="left", padx=12)

    def _build_tab(self, frame: tk.Frame, key: str):
        if key in self.CAT_COLORS:
            c = self.CAT_COLORS[key]
            banner = tk.Frame(frame, bg=c["bg"], pady=10)
            banner.pack(fill="x")
            tk.Label(banner,
                     text=f"{c['icon']}  {key.replace('_', ' ')}",
                     bg=c["bg"], fg=self.WHITE,
                     font=self.FONT_CAT).pack()
        elif key == "restricted":
            banner = tk.Frame(frame, bg="#7f1d1d", pady=10)
            banner.pack(fill="x")
            tk.Label(banner, text="⛔  Restricted (Age > 25)",
                     bg="#7f1d1d", fg="#fca5a5",
                     font=self.FONT_CAT).pack()
        else:
            banner = tk.Frame(frame, bg="#1e293b", pady=10)
            banner.pack(fill="x")
            tk.Label(banner, text="📋  All Students",
                     bg="#1e293b", fg=self.TEXT,
                     font=self.FONT_CAT).pack()

        cols = ("MatNo", "Name", "Age", "Grade", "Category")
        if key != "All Students":
            cols = ("MatNo", "Name", "Age", "Grade")

        style = ttk.Style()
        style.configure(f"{key}.Treeview",
                        background=self.PANEL,
                        foreground=self.TEXT,
                        fieldbackground=self.PANEL,
                        borderwidth=0,
                        rowheight=28,
                        font=self.FONT_TABLE)
        style.configure(f"{key}.Treeview.Heading",
                        background="#0f172a",
                        foreground=self.MUTED,
                        font=("Helvetica Neue", 9, "bold"),
                        relief="flat")
        style.map(f"{key}.Treeview",
                  background=[("selected", "#4f46e5")],
                  foreground=[("selected", self.WHITE)])

        tree_frame = tk.Frame(frame, bg=self.BG)
        tree_frame.pack(fill="both", expand=True, padx=0, pady=0)

        tree = ttk.Treeview(tree_frame, columns=cols,
                             show="headings",
                             style=f"{key}.Treeview")

        widths = {"MatNo": 80, "Name": 200, "Age": 60, "Grade": 70, "Category": 140}
        for col in cols:
            tree.heading(col, text=col)
            tree.column(col, width=widths[col],
                        anchor="center" if col != "Name" else "w")

        vsb = ttk.Scrollbar(tree_frame, orient="vertical", command=tree.yview)
        tree.configure(yscrollcommand=vsb.set)
        tree.pack(side="left", fill="both", expand=True)
        vsb.pack(side="right", fill="y")

        self._trees[key] = tree

        attr_key = key.replace(" ", "_")
        count_var = tk.StringVar(value="")
        tk.Label(frame, textvariable=count_var,
                 bg=self.BG, fg=self.MUTED,
                 font=self.FONT_TINY).pack(pady=4)
        setattr(self, f"_count_{attr_key}", count_var)

        if key == "restricted":
            self.err_var = tk.StringVar(value="")
            err_frame = tk.Frame(frame, bg="#450a0a", padx=16, pady=12)
            err_frame.pack(fill="x", padx=0, pady=(0, 0))
            tk.Label(err_frame, textvariable=self.err_var,
                     bg="#450a0a", fg="#fca5a5",
                     font=self.FONT_ERR,
                     wraplength=720, justify="left").pack(anchor="w")


    def _load_file(self):
        path = filedialog.askopenfilename(
            title="Select SIS CSV file",
            filetypes=[("CSV files", "*.csv"), ("All files", "*.*")]
        )
        if path:
            self.csv_path = path
            self.path_var.set(f"📄  {os.path.basename(path)}")
            self.run_btn.configure(state="normal")
            self.status_var.set(f"File loaded: {path}")


    def _run_classifier(self):
        out_dir = os.path.join(os.path.dirname(self.csv_path), "PAU_Categories")
        self.classifier = SISClassifier(self.csv_path, out_dir)
        try:
            self.classifier.run()
        except Exception as e:
            messagebox.showerror("Error", str(e))
            return

        self._populate_tables()
        self.status_var.set(
            f"✓  Classification complete. CSV files saved to: {out_dir}"
        )

    def _populate_tables(self):
        c = self.classifier

        def _cat_for(s: StudentRecord) -> str:
            age = s.get_age()
            for cat, rule in SISClassifier.CATEGORIES.items():
                if rule(age):
                    return cat.replace("_", " ")
            return "Restricted" if age > 25 else "Boundary"

        tree = self._trees["All Students"]
        tree.delete(*tree.get_children())
        for s in c.get_all_students():
            tree.insert("", "end", values=(
                s.get_mat_no(), s.get_name(),
                s.get_age(), s.get_grade(),
                _cat_for(s)
            ))
        total = len(c.get_all_students())
        getattr(self, "_count_All_Students").set(f"{total} students total")

        for cat in SISClassifier.CATEGORIES:
            tree = self._trees[cat]
            tree.delete(*tree.get_children())
            members = c.get_group(cat)
            for s in members:
                tree.insert("", "end", values=(
                    s.get_mat_no(), s.get_name(),
                    s.get_age(), s.get_grade()
                ))
            key_safe = cat
            getattr(self, f"_count_{key_safe}").set(
                f"{len(members)} student(s) in {cat.replace('_', ' ')}"
            )

        tree = self._trees["restricted"]
        tree.delete(*tree.get_children())
        restricted = [s for s in c.get_all_students() if s.get_age() > 25]
        for s in restricted:
            tree.insert("", "end", values=(
                s.get_mat_no(), s.get_name(),
                s.get_age(), s.get_grade()
            ))
        getattr(self, "_count_restricted").set(
            f"{len(restricted)} restricted student(s)"
        )
        err = c.get_restricted_error()
        self.err_var.set(err if err else "No restricted students found.")


    def _center(self, w: int, h: int):
        self.update_idletasks()
        x = (self.winfo_screenwidth()  - w) // 2
        y = (self.winfo_screenheight() - h) // 2
        self.geometry(f"{w}x{h}+{x}+{y}")


if __name__ == "__main__":
    App().mainloop()
```
